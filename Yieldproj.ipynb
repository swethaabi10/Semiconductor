```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

 #Load the dataset
df = pd.read_csv('signal-data.csv')

#Separate the 'Time' column
time_column = df['Time']

# Replace NaN values with '0'
df = df.fillna(value='0')

# Convert data type of columns to numeric
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Detect columns with all 0 values and delete them
zero_columns = df.columns[(df == 0).all(axis=0)]
df = df.drop(zero_columns, axis=1)

# Drop the existing 'Time' column
df.drop('Time', axis=1, inplace=True)

# Create a new dataframe with the cleaned dataframe
df_Newdata = pd.concat([time_column, df], axis=1)

# Convert 'Time' column to numeric representation
#df_Newdata['Time'] = pd.to_datetime(df_Newdata['Time']).astype('int64') // 10**9
#df_Newdata.select_dtypes(exclude=['object', 'Time']) = df_Newdata.select_dtypes(exclude=['object', 'Time']).apply(pd.to_numeric, errors='coerce')

df_Newdata = df_Newdata.select_dtypes(include=['int64', 'float64']).apply(pd.to_numeric, errors='coerce')
# Segregate predictors and target attributes
X = df_Newdata.drop('Pass/Fail', axis=1)
y = df_Newdata['Pass/Fail']

# Check and fix target imbalance
smote = SMOTE()
X_res, y_res = smote.fit_resample(X, y)

# Perform train-test split and standardize data
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Perform PCA on the scaled data
pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
```

```python
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Initialize the models
models = {
    'SVM': SVC(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Naive Bayes': GaussianNB()
}

# Dictionary to store accuracy scores
accuracy_results = {}

# 2. Train and evaluate each model
for model_name, model_instance in models.items():
    # Train the model
    model_instance.fit(X_train_scaled, y_train)
    
    # Make predictions
    y_pred = model_instance.predict(X_test_scaled)
    
    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)
    accuracy_results[model_name] = accuracy
    
    # Print individual results
    print(f"--- {model_name} ---")
    print(f"Accuracy: {accuracy:.4f}\n")

# 3. Compare and find the best model
best_model_name = max(accuracy_results, key=accuracy_results.get)
best_accuracy = accuracy_results[best_model_name]

print("==================================================")
print(f"🏆 BEST MODEL: {best_model_name} with {best_accuracy:.4f} accuracy")
print("==================================================")

# 4. Save the best model to the 'model' variable so it works with your feature importance chart (if it's Random Forest)
model = models[best_model_name]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import joblib

# 1. Print detailed metrics for the winning SVM model
print("Classification Report for Best Model (SVM):\n")
y_pred_best = model.predict(X_test_scaled)
print(classification_report(y_test, y_pred_best))

# 2. Visualize the Confusion Matrix for your portfolio
cm = confusion_matrix(y_test, y_pred_best)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Pass', 'Fail'], yticklabels=['Pass', 'Fail'])
plt.title('Confusion Matrix - SVM Yield Prediction')
plt.xlabel('Predicted Outcome')
plt.ylabel('Actual Outcome')
plt.show()

# 3. Save the model
joblib.dump(model, 'svm_yield_model.joblib')
print("SVM model successfully saved as a .joblib file!")
```
